## 5. 연령대별 / 성별 유동인구 합계 분석
- 남성과 여성 컬럼을 분리하고, 블록별 총 유동인구를 파생변수로 생성합니다.

```python
# 남성 및 여성 연령대 컬럼 목록 분리
man_cols = [c for c in flow_age_2507.columns if c.startswith('MAN_FLOW_POP_CNT_')]
wman_cols = [c for c in flow_age_2507.columns if c.startswith('WMAN_FLOW_POP_CNT_')]

# 블록별 총 남성, 총 여성, 총 유동인구 합계 파생변수 생성
flow_age_2507['TOTAL_MAN_FLOW'] = flow_age_2507[man_cols].sum(axis=1)
flow_age_2507['TOTAL_WMAN_FLOW'] = flow_age_2507[wman_cols].sum(axis=1)
flow_age_2507['TOTAL_FLOW'] = flow_age_2507['TOTAL_MAN_FLOW'] + flow_age_2507['TOTAL_WMAN_FLOW']

# 총 유동인구가 가장 많은 상위 5개 블록 확인
top_blocks = flow_age_2507.sort_values(by='TOTAL_FLOW', ascending=False)
top_blocks[['STD_YM', 'BLOCK_CD', 'X_COORD', 'Y_COORD', 'TOTAL_MAN_FLOW', 'TOTAL_WMAN_FLOW', 'TOTAL_FLOW']].head()
```

```python
import pandas as pd

table_def_path = '테이블정의서.xlsx'
df_def = pd.read_excel(table_def_path, sheet_name=0)

# '컬럼ID'라는 텍스트가 들어있는 행의 인덱스를 동적으로 탐색
header_row_idx = None
for idx, row in df_def.iterrows():
    if '컬럼ID' in row.values:
        header_row_idx = idx
        break

if header_row_idx is not None:
    # 해당 행을 헤더로 다시 설정
    df_def.columns = df_def.iloc[header_row_idx]
    df_def = df_def.iloc[header_row_idx + 1:].reset_index(drop=True)

    # 딕셔너리 생성
    mapping_dict = dict(zip(df_def['컬럼ID'].dropna().str.strip(), df_def['컬럼명'].dropna().str.strip()))
    print("✅ 매핑 사전 추출 성공! (총 개수:", len(mapping_dict), ")")
else:
    print("❌ '컬럼ID' 헤더를 찾지 못했습니다.")
```

## 6. 연령대별 남녀 유동인구 총합 비교 테이블

```python
age_labels = ['10대', '20대', '30대', '40대', '50대', '60대이상']
man_totals = [flow_age_2507[c].sum() for c in man_cols]
wman_totals = [flow_age_2507[c].sum() for c in wman_cols]

summary_df = pd.DataFrame({
    '연령대': age_labels,
    '남성 유동인구': [round(x, 1) for x in man_totals],
    '여성 유동인구': [round(x, 1) for x in wman_totals]
})
summary_df['전체 유동인구'] = summary_df['남성 유동인구'] + summary_df['여성 유동인구']
summary_df['남성 비율(%)'] = round(summary_df['남성 유동인구'] / summary_df['전체 유동인구'] * 100, 1)
summary_df['여성 비율(%)'] = round(summary_df['여성 유동인구'] / summary_df['전체 유동인구'] * 100, 1)

display(summary_df)
```

```python
import pandas as pd

# 1. 테이블 정의서에서 컬럼 ID -> 한글 컬럼명 매핑 사전 추출
table_def_path = '테이블정의서.xlsx'
df_def = pd.read_excel(table_def_path, sheet_name=0)

mapping_dict = {}
for idx, row in df_def.iterrows():
    col_id = row['Unnamed: 1']
    col_name = row['Unnamed: 2']
    if pd.notna(col_id) and pd.notna(col_name):
        col_id_str = str(col_id).strip()
        col_name_str = str(col_name).strip()
        if col_id_str != '컬럼ID':
            mapping_dict[col_id_str] = col_name_str
```

```python
# 2. 뭉쳐진 원본 데이터 불러오기 및 '|' 구분자로 분해
data_path = 'flow_age_pop_202507.xlsx'
df_raw = pd.read_excel(data_path)

# 헤더 분리 및 행 데이터 분해
col_string = df_raw.columns[0]
raw_col_names = col_string.split('|')

df_split = df_raw[col_string].astype(str).str.split('|', expand=True)
df_split.columns = raw_col_names

# 3. 한글 컬럼명 매핑 적용
df_mapped = df_split.rename(columns=mapping_dict)

# 수치형 컬럼 타입 변환
for col in df_mapped.columns:
    try:
        df_mapped[col] = pd.to_numeric(df_mapped[col])
    except (ValueError, TypeError):
        pass


# 4. CSV 파일 저장 (한글 깨짐 방지 utf-8-sig 적용)
output_csv_path = 'flow_age_pop_202507_mapped.csv'
df_mapped.to_csv(output_csv_path, index=False, encoding='utf-8-sig')
print("작업이 성공적으로 완료되었습니다.")
```
